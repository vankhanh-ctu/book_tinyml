# Numpy

```python
# Import NumPy for numerical operations and array handling
import numpy as np
```

```python
# Create a 1-D array with integers from 0 to 11
x = np.arange(12)
# Evaluate / display the array
x
```

```python
# Return the number of elements in the array x
len(x)
```

```python
# Show the shape (dimensions) of the array x
x.shape
```

```python
# Reshape the 1-D array x into a 3x4 2-D array and assign to X
X = x.reshape(3, 4)
# Display the reshaped matrix
X
```

```python
# Create an array of zeros with shape (3, 3, 4)
# This could represent a batch of 3 matrices of size 3x4 filled with zeros
matrix_0 = np.zeros((3, 3, 4))
# Display the zero-initialized tensor
matrix_0
```

```python
# Create an array of ones with shape (3, 3, 4)
# Useful for testing or as a neutral additive tensor
matrix_1 = np.ones((3, 3, 4))
# Display the ones-initialized tensor
matrix_1
```

```python
# Create a 3x3x4 array populated with samples from a normal distribution
# loc=0.0 is the mean, scale=1.0 is the standard deviation
matrix_random = np.random.normal(loc=0.0, scale=1.0, size=(3, 3, 4))
# Display the randomly sampled tensor
matrix_random
```

```python
# Demonstrate indexing and slicing on the 2-D array X
# X[-1] accesses the last row, X[1:3] slices rows 1 and 2 (end exclusive)
X[-1], X[1:3]
```

```python
# Modify a single element of X using zero-based indexing
# Set the element at row index 2, column index 3 to the value 20
X[2, 3] = 20
# Display the updated matrix to confirm the change
X
```

```python
# Broadcast the scalar value 1 to the first two rows of X
# This sets all elements in rows 0 and 1 to 1
X[:2, :] = 1
# Display X after broadcasting assignment
X
```

```python
# Create two 1-D arrays and demonstrate element-wise arithmetic
x = np.array([1, 2, 4, 8])
y = np.array([2, 2, 2, 2])
# Compute and return (add, subtract, multiply, divide, power) element-wise
x + y, x - y, x * y, x / y, x ** y
```

```python
# Create two 2-D arrays and show concatenation along both axes
# X is 3x4 with values 0..11, Y is a 3x4 example matrix
X = np.arange(12).reshape((3,4))
Y = np.array([[2.0, 1, 4, 3], [1, 2, 3, 4], [4, 3, 2, 1]])
# Concatenate vertically (axis=0) and horizontally (axis=1)
np.concatenate((X, Y), axis=0), np.concatenate((X, Y), axis=1)
```

# Matplotlib

```python
# Import matplotlib for data visualization
import matplotlib.pyplot as plt
from matplotlib import rcParams

# Configure font settings for better readability
rcParams['font.family'] = 'serif'
rcParams['font.serif'] = ['Times New Roman']

# Define sample data points
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# Plot the data as a line graph with styling
plt.plot(x, y, color='blue', marker='o', linestyle='--', label='y = 2x')

# Set chart title and axis labels with custom font size
plt.title('Simple Line Chart with Matplotlib', fontsize=16)
plt.xlabel('X Axis', fontsize=16)
plt.ylabel('Y Axis', fontsize=16)
plt.xticks(fontsize=16)
plt.yticks(fontsize=16)

# Add legend to identify the plotted line
plt.legend(fontsize=16)

# Render the plot
plt.show()
```

# Pandas

```python
# Import pandas for data manipulation and analysis
import pandas as pd

# Create a simple DataFrame with student information
data = {
    'Tên': ['An', 'Bình', 'Chi'],      # Names
    'Tuổi': [23, 25, 22],               # Ages
    'Điểm': [8.5, 7.8, 9.0]             # Scores
}

# Convert dictionary to DataFrame
df = pd.DataFrame(data)

# Display the original data
print("Dữ liệu ban đầu:")
print(df)

# Calculate the average score
diem_tb = df['Điểm'].mean()
print("\nĐiểm trung bình:", diem_tb)

# Filter students with scores above 8
sv_gioi = df[df['Điểm'] > 8]
print("\nSinh viên có điểm trên 8:")
print(sv_gioi)
```
